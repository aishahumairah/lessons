### Exercise 6

Write a function `clean_text(value, case='title')` that:
1. Converts the value to a string
2. Strips leading/trailing whitespace
3. Applies the casing specified by `case` (`'title'`, `'lower'`, or `'upper'`)
4. Returns the cleaned string

Test it with `'  ELECTRONICS  '` in all three modes.

````python
# Your answer here
def clean_text(value, case='title'): 
    cleaned = str(value).strip() 
    if case == 'title':
        return cleaned.title()
    elif case == 'lower':
        return cleaned.title()
    else: return cleaned.upper()
test = ' ELECTRONICS '
print (clean_text(test, case='title'))
    
````

### Exercise 7

Write a function `safe_float(value)` that converts a value to float and returns `None` (not an error) if conversion fails.

Test it with: `'$350.00'`, `'N/A'`, `''`, `75.0`

Hint: use `try/except`.

````python
# Your answer here
def safe_float(value):
   try: 
       cleaned = str(value).replace('$','').strip()
       return float(cleaned)
   except (ValueError, TypeError):
        return None
print(safe_float(''))
````

### Exercise 8

Write a function `validate_order(row)` that accepts a dict and returns `(True, 'OK')` if valid, or `(False, reason)` if not.

Validation rules:
- `quantity` must be greater than 0
- `unit_price` must be 0 or greater
- `total` must equal `quantity * unit_price` (within 0.01)

Test it with a valid row and two different invalid rows.

````python
# Your answer here
def validate_order(row):
    quantity = row['quantity']
    unit_price = row['unit_price']
    total = row['total']
    if quantity <=0:
        return (False, 'must be greater than 0')
    if unit_price <0:
        return (False, 'must be 0 or greater')
    expected_total = quantity * unit_price 
    if (total - expected_total) >0.01:
        return (False, 'total must = expected_total')
    return (True, 'OK')
valid_row = {'quantity':4,'unit_price':-10.0, 'total':40.0}
print (validate_order(valid_row)) 
````

### Exercise 9

Using pathlib, write a function `list_csv_files(folder)` that:
1. Takes a folder path as a string or Path
2. Returns a sorted list of all `.csv` filenames (not full paths — just names) in that folder

Test it against `'../data/raw'`.

Expected output: `['sales_q1.csv', 'sales_q2.csv', 'sales_q3.csv']`

````python
# Your answer here
from pathlib import Path
def list_csv_files(folder):
    cleaned = Path(folder)
    csv_files = [p.name for p in cleaned.glob('*.csv')]
    return sorted(csv_files)
    
print(list_csv_files('../data/raw'))
````

### Exercise 16

Using `q2` (or a copy of it), clean the `category` column:

2. Convert to title case
3. Replace `'Electrnics'` with `'Electronics'` and `'Home & Kithen'` with `'Home & Kitchen'`

Print the value counts after each step to show the change.

### Exercise 17

Load `sales_q1.csv`. The `unit_price` column has some values prefixed with `$`. Clean it:
1. Remove the `$`
2. Convert to float using `pd.to_numeric(..., errors='coerce')`
3. Print the dtype before and after to confirm it changed

````python
# Your answer here
q1_copy = q1.copy()
q1_copy['unit_price'] = q1_copy['unit_price'].astype(str).str.replace('$','', regex=False).str.strip()
print(q1_copy['unit_price'].dtype)
q1_copy['unit_price'] = pd.to_numeric(q1_copy['unit_price'], errors='coerce')
print(q1_copy['unit_price'].dtype)
````

### Exercise 20

Write a function `read_all_quarters(folder)` that:
1. Accepts a folder path
2. Reads every `.csv` file in that folder into a separate DataFrame
3. Returns a list of DataFrames

Test it against `'../data/raw'` and print the shape of each DataFrame it returns.

````python
# Your answer here
import pandas as pd
from pathlib import Path
def read_all_quarters(folder):
    cleaned_new = Path(folder)
    csv_paths = sorted(cleaned_new.glob('*.csv'))
    csv_files = [pd.read_csv(p) for p in csv_paths]
    return csv_files
dataframe = read_all_quarters('../data/raw')
for d in dataframe: 
    print(d.shape)
````

### Exercise 21

Write a function `revenue_by_category(df)` that:
1. Takes a clean DataFrame (already has numeric `total` and normalised `category`)
2. Groups by `category`
3. Returns a DataFrame with columns `orders` (count) and `revenue` (sum), sorted descending by revenue

Test it on `sales_q1.csv` after applying basic cleaning.

````python
# Your answer here
import pandas as pd

def revenue_by_category(df):
    
    return (
        df.groupby('category')
        .agg(
            orders=('order_id', 'count'), 
            revenue=('total', 'sum')
        )
        .sort_values(by='revenue', ascending=False)
    )
print(revenue_by_category(q1_copy))
````
